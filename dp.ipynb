```python
def frogJump(heights):
    total_energy = 0
    i = 0
    while i < len(heights) - 2:
        if abs(heights[i] - heights[i+1]) < abs(heights[i] - heights[i+2]):
            total_energy += abs(heights[i] - heights[i+1])
            i += 1
        else:
            total_energy += abs(heights[i] - heights[i+2])
            i += 2
        print(total_energy, i)
    if i != len(heights) - 1:
        total_energy += abs(heights[i] - heights[len(heights) - 1]) 
    return total_energy

heights = [7, 5, 1, 2, 6]
frogJump(heights)   
```

```python
class Solution:
    def minCost(self, heights, k):
        n= len(heights)
        dp = [-1] * n
        dp[0] = 0
        for ind in range(1,n):
            min_dist = float("inf")
            for dec in range(1,k+1):
                if ind - dec >= 0:
                    # print(abs(heights[ind] - heights[ind-dec]))
                    # print(ind, ind - dec)
                    min_dist = min(min_dist, dp[ind-dec] + (abs(heights[ind] - heights[ind-dec])))
            dp[ind] = min_dist
            
        return dp[n - 1]

s = Solution()
s.minCost(heights = [15, 4, 1, 14, 15], k = 3)
```

```python
nums = [2, 1, 4, 9, 5, 7, 1, 9, 20, 10]

def findMax(nums):
    n = len(nums)
    dp = [-1] * n
    max_sum = float("-inf")
    
    for i in range(n):
        if i <= 1:
            dp[i] = nums[i]
            max_sum = max(max_sum, dp[i])
            continue
        if i <= 2:
            dp[i] = dp[i-2] + nums[i]
            max_sum = max(max_sum, dp[i])
            continue         
        first =  dp[i-2] + nums[i]
        second = dp[i-3] + nums[i]
        dp[i] = max(first,second)
        max_sum = max(max_sum, dp[i])
    return max_sum

findMax(nums)
```

```python
class Solution:
    def circular(self,nums):
        n = len(nums)
        return max(self.rob(nums[0:n-1]), self.rob(nums[1:n])) 
    def rob(self, nums) -> int:
        n = len(nums)
        dp = [-1] * n
        max_sum = float("-inf")
        prev1, prev2, prev3 = 0,0,0
        for i in range(n):
            if i <= 1:
                curr = nums[i]
                max_sum = max(max_sum, curr)
                prev3 = prev2
                prev2 = prev1
                prev1 = curr
                continue
            if i <= 2:
                curr = prev2 + nums[i]
                prev3 = prev2
                prev2 = prev1
                prev1 = curr
                max_sum = max(max_sum, curr)
                continue         
            first =  prev2 + nums[i]
            second = prev3 + nums[i]
            curr = max(first,second)
            prev3 = prev2
            prev2 = prev1
            prev1 = curr
            max_sum = max(max_sum, curr)
        return max_sum

money = [2, 1, 4, 9]
s = Solution()
s.circular(money)
```
