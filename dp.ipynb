# Unbounded Knapsack (DP-23)

Problem Statement: A thief wants to rob a store. He is carrying a bag of capacity W. The store has ‘n’ items of infinite supply. Its weight is given by the ‘wt’ array and its value by the ‘val’ array. He can either include an item in its knapsack or exclude it but can’t partially have it as a fraction. We need to find the maximum value of items that the thief can steal. He can take a single item any number of times he wants and put it in his knapsack .

```python
def unboundKnapsack(weights, values, limit):
    n= len(weights)
    knapsack = [[] for _ in range(n)]
    for i in range(n):
        knapsack[i].append(weights[i])
        knapsack[i].append(values[i])
    
    prev = [0] * (limit+1)
    
    for i in range(n):
        cur = [0] * (limit+1)
        for j in range(limit+1):
            not_taken = prev[j]
            take = float('-inf')
            if knapsack[i][0] <= j:
                take = knapsack[i][1] + cur[j - knapsack[i][0]]
            cur[j] = max(not_taken, take)
        prev = cur
    return prev

W = 8
wt = [2, 4, 6]
val = [5, 11, 13]
unboundKnapsack(wt, val, W)
```

```python
class Solution:
    # Function to find the minimum multiplication cost
    def matrixChainOrder(self, arr, i, j,dp):
        # Base case: if only one matrix is present, no multiplication needed
        if i == j:
            return 0

        # Initialize minimum cost as very large
        min_cost = float("inf")
        if dp[i][j] != -1:
            return dp[i][j]
        # Try all possible partitions
        for k in range(i, j):
            # Cost of multiplying matrices from i to k
            cost1 = self.matrixChainOrder(arr, i, k,dp)

            # Cost of multiplying matrices from k+1 to j
            cost2 = self.matrixChainOrder(arr, k + 1, j,dp)

            # Cost of multiplying the two resulting matrices
            costMultiply = arr[i - 1] * arr[k] * arr[j]

            # Total cost for this partition
            total = cost1 + cost2 + costMultiply

            # Update minimum
            min_cost = min(min_cost, total)
        dp[i][j] = min_cost
        return min_cost


# Driver code
if __name__ == "__main__":
    sol = Solution()
    arr = [10, 20, 30, 40]
    n = len(arr)
    dp = [[-1] * (len(arr)) for _ in range(len(arr))]
    print("Minimum number of multiplications is:",
    sol.matrixChainOrder(arr, 1, n - 1,dp))
    print(dp)


```

```python
class Solution:
    def evaluateExp(self, exp):
        n = len(exp)
        dp = [[[-1 for _ in range(2)] for _ in range(n+1)] for _ in range(n+1)]
        return self.memorization(0, n-1, 1, exp, dp),dp
    def memorization(self, i, j, isTrue, exp, dp):
        if i > j:
            return 0
        
        if i == j:
            if isTrue == 1:
                return 1 if exp[i] == 'T' else 0
            else:
                return 1 if exp[i] == 'F' else 0
        
        if dp[i][j][isTrue] != -1:
            return dp[i][j][isTrue]
        
        ways = 0
        
        for ind in range(i+1,j,2):
            LT = self.memorization(i, ind-1, 1, exp, dp)
            LF = self.memorization(i, ind-1, 0, exp, dp)
            RT = self.memorization(ind+1, j, 1, exp, dp)
            RF = self.memorization(ind+1, j, 0, exp, dp)
            
            if exp[ind] == '&':
                if isTrue == 1:
                    ways = LT * RT
                else:
                    ways = LT * RF + RT * LF + RF * LF
            elif exp[ind] == '|':
                if isTrue == 1:
                    ways +=  LT * RF + RT * LF + RT * LT
                else:
                    ways = LF * RF
            else:
                if isTrue == 1:
                    ways += LF * RT + RF * LT
                else:
                    ways += LF * RF + RT * LT
        dp[i][j][isTrue] = ways % 1000000007  # Store the result in the DP table
        return dp[i][j][isTrue]

if __name__ == "__main__":
    exp = "F|T^F"  # Example expression
    
    # Evaluate the expression and find the number of ways to get the result as True
    sol = Solution()
    ways = sol.evaluateExp(exp)
    
    # Output the result
    print("The total number of ways:", ways)
```

```python
def memoriation(i,j,s,dp):
    if i >j:
        return 0
    if dp[i][j] != -1:
        return dp[i][j]
    if s[i:j+1] == s[i:j+1:-1]:
        dp[i][j] = 0
        return dp[i][j]
        
    min_cuts = float("inf")
    for k in range(i,j+1):
        cuts = 1 + memoriation(i,k-1,s,dp) + memoriation(k+1,j,s,dp)
        min_cuts = min(min_cuts, cuts)
    dp[i][j] = min_cuts
    return min_cuts
def palinfromeCuts(s):
    n = len(s)
    dp = [[-1] * n for _ in range(n)]
    return memoriation(0,n-1,s,dp)

s = 'bababcbadcede'
palinfromeCuts(s)
    
```

```python
class Solution:
    # Check if substring s[start...end] is palindrome
    def isPalindrome(self, s: str, start: int, end: int) -> bool:
        while start < end:
            if s[start] != s[end]:
                return False
            start += 1
            end -= 1
        return True

    # Recursive helper with memoization to find min cuts from index 'start'
    def minCutsHelper(self, s: str, start: int, memo: list[int]) -> int:
        n = len(s)

        # If reached end or substring is palindrome, no cut needed
        if start == n or self.isPalindrome(s, start, n - 1):
            return 0

        # Return stored result if already computed
        if memo[start] != -1:
            return memo[start]

        minCuts = float('inf')

        # Try all possible partitions
        for end in range(start, n):
            if self.isPalindrome(s, start, end):
                # 1 cut plus cuts for the remaining substring
                cuts = 1 + self.minCutsHelper(s, end + 1, memo)
                minCuts = min(minCuts, cuts)

        # Store and return minimum cuts needed from this start
        memo[start] = minCuts
        return minCuts

    # Public function to get minimum cuts for palindrome partitioning
    def minCut(self, s: str) -> int:
        n = len(s)
        memo = [-1] * n
        return self.minCutsHelper(s, 0, memo)


if __name__ == "__main__":
    s = 'bababcbadcede'

    sol = Solution()
    print("Minimum cuts needed:", sol.minCut(s))
```

```python
def max_product(arr):
    n = len(arr)
    dp = [[-1] * 2 for i in range(n+1)]
    if arr[0] <= 0:
        dp[0][1] = 1
        dp[0][0] = arr[0]
        first_neg = True
        
    else:
        dp[0][1] = arr[0]
        dp[0][0] = arr[0]
        first_neg = False
    max_num = float("-inf")
    for i in range(1,n):
        if arr[i] == 0:
            dp[i][0] = 1
            dp[i][0] = 1
            max_num = max(0,max_num)
            first_neg = False
            continue
        if arr[i] < 0 and first_neg == False:
            dp[i][0] = dp[i-1][0] * arr[i]
            dp[i][1] = 1
            max_num = max(dp[i][0], max_num)
            first_neg = True
            continue
        else:
            dp[i][0] = dp[i-1][0] * arr[i]
            dp[i][1] = dp[i-1][1] * arr[i]
        max_num = max(dp[i][0],dp[i][1],max_num)
    return max_num

max_product([-5,0,-6,-9])
```

```python
nums = [5,7,-6,-7,4,-102,0,-43,345,-90]
def maxProduct(nums):
    res = max(nums)
    cur_max = cur_min = 1

    for n in nums:
        temp = cur_max * n
        cur_max = max(temp, cur_min * n, n)
        cur_min = min(temp, cur_min * n, n)

        res = max(res, cur_max)
    
    return res
maxProduct(nums)
```
