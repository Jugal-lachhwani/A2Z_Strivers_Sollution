```python
arr = [2,2,1,1,3,3,5, 8,8,2]

def SingleNoIII(arr):
    xor = 0
    for i in arr:
        xor ^= i
    
    rightmost_diff = (xor & xor - 1) ^ xor
    
    xor0 = 0
    xor1 = 0
    for i in arr:
        if i & rightmost_diff == 0:
            xor0 ^= i
        else:
            xor1 ^= i
            
    return (xor1,xor0) if xor1 < xor0 else (xor0,xor1)

SingleNoIII(arr)
```
